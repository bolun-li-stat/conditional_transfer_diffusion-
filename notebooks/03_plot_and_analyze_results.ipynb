```python
from pathlib import Path
from plot_results import load_metrics, aggregate, plot_metric_curves, plot_mismatch_differences, plot_summary_bars, plot_training_losses, plot_pca_samples, textual_summary
from utils import ensure_dir

results_dir = Path('results')
figure_dir = ensure_dir(results_dir / 'figures')
df = load_metrics(results_dir)
print(df.shape)
df.head()
```

```python
agg = aggregate(df)
agg.to_csv(results_dir / 'aggregated_metrics.csv', index=False)
agg.head()
```

```python
plot_metric_curves(agg, figure_dir)
plot_mismatch_differences(df, figure_dir)
plot_summary_bars(agg, figure_dir)
plot_training_losses(results_dir, figure_dir)
# PCA plots are generated only if sample files were saved with save_samples=True.
plot_pca_samples(results_dir, figure_dir)
print('Saved figures to', figure_dir)
```
