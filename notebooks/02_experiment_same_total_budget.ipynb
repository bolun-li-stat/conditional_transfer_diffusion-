```python
import torch
print('torch.cuda.is_available():', torch.cuda.is_available())
if torch.cuda.is_available():
    print('GPU:', torch.cuda.get_device_name(0))
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
print('Using device:', DEVICE)
```

```python
from config import default_experiment2_configs
from train import run_single_setting

configs = default_experiment2_configs(seeds=[0, 1, 2])
for cfg in configs:
    cfg.training.device = DEVICE
    cfg.training.training_steps = 20_000
    cfg.training.resume_from_checkpoint = True
    cfg.results_dir = Path('results')
    cfg.evaluation.save_samples = False

print(f'Prepared {len(configs)} settings.')
```

```python
for i, cfg in enumerate(configs, start=1):
    print(f'Running setting {i}/{len(configs)}:', cfg.data.covariance_scenario, cfg.data.rho, cfg.data.mismatch_level, cfg.n, 'seed', cfg.seed)
    run_single_setting(cfg, force=False)
```
